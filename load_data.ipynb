```python
dir='/content/drive/MyDrive/CP_DS/DS'
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```python
img_height, img_width = 75, 75
batch_size = 32

ds = tf.keras.utils.image_dataset_from_directory(
    dir,
    labels='inferred',
    label_mode='int',
    image_size=(img_height, img_width),
    interpolation='nearest',
    batch_size=batch_size,
    shuffle=True
)
```

```python
ds.class_names
```

```python
print('Number of batch',tf.data.experimental.cardinality(ds).numpy())
```

```python
for image_batch, labels_batch in ds.take(1):
  print(image_batch.shape)
  print(labels_batch.shape)
  print(labels_batch)
  plt.figure(figsize=(10,10))
  for i in range(batch_size):
    plt.subplot(4,8,i+1)
    plt.imshow(image_batch[i])
    plt.title(ds.class_names[labels_batch[i]])
    plt.axis('off')
```

```python
img=[]
label=[]
for image_batch,label_batch in ds:
      img.append(image_batch)
      label.append(label_batch)
      data_input=np.concatenate(img)
      data_target=np.concatenate(label)
print(data_input.shape)
print(data_target.shape)
```
